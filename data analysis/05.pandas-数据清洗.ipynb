```python
import pandas as pd

pd.set_option("display.min_rows", 30)

df = pd.read_csv('data/movies_list_02.csv')

# 数据清洗
# # 缺失值处理
# df.isnull() # 查看缺失值

# # 1。删除缺失值
# df = df.dropna() # 删除包含缺失值的行
# df = df.dropna(axis=1)

# 2。填充缺失值
df = df.fillna('*')
df
# df.ffill() # 填充缺失值，填充上一行数据
# df.bfill() # 填充缺失值，填充下一行数据
```

```python
# 重复值
df.duplicated() # 所有列都重复
df.duplicated(subset=['语言'])

# 删除重复值
df.drop_duplicates(subset=['语言'],keep='last') #keep = 'last' 删除重复值，保留最后一个; keep = 'first' 删除重复值，保留第一个
```

```python
# 异常值的处理
# 查看异常数据
df[df['评分'] > 100]

# 删除异常数据
df.drop(df[df['评分'] > 100].index)

# 修复异常值
df['评分'] = df['评分'].abs() # 绝对值

```

```python
# 数据格式处理
df['上映时间'] = df['上映时间'].str.replace('/','-')
df
```
