```python
import pandas as pd

df = pd.read_csv('data/movies_list_02.csv',nrows=30)

# 排序
df.sort_values('时长',ascending=False)

#
df.sort_values(['评分','时长'],ascending=[False,True])
df.sort_values(['评分','时长'],ascending=True)
```

```python
# 把“动画,动作,冒险,科幻”拆成列表，再展开
movies = df.assign(
    类型=df["类型"].str.split(",")
).explode("类型")

# 清除可能存在的空格
movies["类型"] = movies["类型"].str.strip()

movies[["电影名", "类型", "评分", "时长"]].head(10)

movies.groupby("类型").count()

```

```python
# 为了正确计算最早上映时间
movies["上映时间"] = pd.to_datetime(movies["上映时间"], errors="coerce")

genre_summary = (
    movies.groupby("类型", as_index=False)
    .agg(
        电影数量=("电影名", "nunique"),
        平均评分=("评分", "mean"),
        最高评分=("评分", "max"),
        平均时长=("时长", "mean"),
        最早上映=("上映时间", "min"),
        最晚上映=("上映时间", "max"),
    )
    .sort_values("电影数量", ascending=False)
)

genre_summary["平均评分"] = genre_summary["平均评分"].round(2)
genre_summary["平均时长"] = genre_summary["平均时长"].round(1)

genre_summary
```
