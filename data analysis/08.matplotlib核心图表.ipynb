```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt
from IPython.core.pylabtools import figsize

# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

# 创建子图
# figure：画布对象 , axes :子图数组（里面存放的是Axes对象）
figure , axes = plt.subplots(nrows = 1 , ncols = 2 , figsize = (20,6) , dpi = 100) # nrows: 行，ncols: 列

# 图一：柱状图
countries = ['美国', '中国', '英国', '日本', '德国','沙特','伊朗']
values = [36, 28, 20, 16, 12, 79, 51]

axes1:Axes = axes[0]
axes1.bar(countries, values,width=0.6,color = 'g')
axes1.set_title('世界石油储备',fontsize = 18 )
axes1.set_xlabel("国家",loc = 'right')
axes1.set_ylabel('石油储备（亿吨）',loc = 'top')

axes1.grid(linestyle = '--',alpha = 0.6) # 网格线

# 图二：饼状图（世界人口） ---> 擅长比例分析 ---> pie方法
countries2 = ['美国', '中国', '英国', '日本', '德国','沙特','伊朗']
values = [4.1,14.6 , 2, 1.1, 1.2, 7.9, 5.1]

axes2:Axes = axes[1]
axes2.pie(values, labels=countries2, autopct='%1.1f%%')
axes2.set_title('世界人口比例',fontsize = 18)
axes2.legend(loc = 'lower center',ncol = 5,bbox_to_anchor = (0.5,-0.1)) # 显示图例， ncol:每行显示个数 ，bbox_to_anchor: 设置图例位置 （0.5：控制水平方位 -0.1：控制垂直方位）

# 保存图片
plt.savefig('data/01.png')
plt.show()

```
