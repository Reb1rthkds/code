```python
# 数据过滤 ---> df[filter]
# 1.获取评分 > 90
# df[df['评分'] > 90]

# 2.获取类别为 科幻或者动作 的电影 ？？？

# 3.获取评分在80-90之间 ---> 默认包含边界
df[df['评分'].between(80,90)]

# 4.获取时长 > 120 并且评分 > 85
df[(df['评分'] >= 85) & (df['时长'] >= 120)]

# 5.获取年份2025/2026 并且 评分在80-99之间
df[(df['年份'].isin([2025,2026])) & (df['评分'].between(80,99))]
```
