```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.spatial import Voronoi, voronoi_plot_2d
# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']


rng = np.random.default_rng(42)

points = rng.random((15,2)) * 10
# x,y = points[:,0], points[:,1]
# plt.plot(x,y,linestyle = 'None',marker='o',markersize = 5)
# print(points)

vor = Voronoi(points)

figure, axes = plt.subplots(figsize=(7, 7))

voronoi_plot_2d(
    vor,
    ax=axes,
    show_points=True,
    points_colors = 'green',
    show_vertices=True,
    line_colors="black",
    line_width=1.2,
    point_size=10
)
axes.set_xticks([i for i in range(0,110)])
axes.set_yticks([i for i in range(0,110)])
axes.set_title('Voronoi图')
axes.set_xlim(0, 20)
axes.set_ylim(0, 20)
axes.set_aspect("equal")
plt.show()
```
